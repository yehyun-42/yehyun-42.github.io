```python
import csv
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("C:/Mydisk/workspaces/SQL_project/Steam_FINAL.csv")
df.head()
```

```python
df.info()
```

```python
need_columns=['app_id', 'name', 'release_date', 'price_usd', 'review_score_pct', 'total_reviews', 'estimated_owners', 
              'steam_official_genres', 'steam_developer']
df_need=df[need_columns].copy()
df_need.head()
```

```python
df_need.isna().sum()
```

```python
df_need=df_need.dropna()
```

```python
df_need.isna().sum()
```

```python
genres_table_column=['app_id', 'name', 'steam_official_genres']
df_genres=df[genres_table_column].copy()
df_genres.head()
```

```python
df_genres['steam_official_genres'] = (
    df_genres['steam_official_genres']
    .astype(str)
    .str.replace(r"[\[\]'\"]", "", regex=True))
df_genres['steam_official_genres']=df_genres['steam_official_genres'].str.split(',')
df_genres=df_genres.explode('steam_official_genres')
df_genres['steam_official_genres']=df_genres['steam_official_genres'].astype(str).str.strip()

df_genres.head(10)
```

```python
final_columns=['app_id', 'name', 'release_date', 'price_usd', 'review_score_pct', 'total_reviews', 'estimated_owners', 'steam_developer']
df_final=df_need[final_columns].copy()
df_final.head()
```

```python
df_final['steam_developer'] = (
    df_final['steam_developer']
    .astype(str)
    .str.replace(r"[\[\]'\"]", "", regex=True))
df_final.head()
```

```python
df_final['review_score_negative']=100-df_final['review_score_pct']
df_final.head()
```

```python
df_final=df_final[df_final.total_reviews != 0]
df_final.head(10)
```

```python
df_final['name'] = df_final['name'].astype(str).str.replace(r"[^a-zA-Z0-9\s]", "", regex=True)
df_final['steam_developer'] = df_final['steam_developer'].astype(str).str.replace(r"[^a-zA-Z0-9\s]", "", regex=True)
df_final.head(20)
```

```python
df_final.to_csv('game_table.csv', index=False)
df_genres.to_csv("genres_table.csv", index=False)
```
