```python
def is_clean_name(text):
    if pd.isna(text):
        return False
    try:
        str(text).encode('utf-8').decode('utf-8')
        import re
        if re.search(r'[\u8e00-\u9fff\u3000-\u3fff]+', str(text)):
            return False
        return True
    except:
        return False

df_need = df_need[df_need['Name'].apply(is_clean_name)]
df_need.head()
```

```python
df_need['AppID']=pd.to_numeric(df_need['AppID'], errors='coerce')
df_need['Positive']=pd.to_numeric(df_need['Positive'], errors='coerce')
df_need['Negative']=pd.to_numeric(df_need['Negative'], errors='coerce')
df_need.info()
```

```python
import numpy as np
```

```python
def drop_outliers_zscore(df, column, threshold):
    z_scores=(df[column]-df[column].mean()) / df[column].std()
    df_drop_outliers=df[abs(z_scores) <= threshold]
    return df_drop_outliers
```

```python
df_last=drop_outliers_zscore(df_need, 'Negative', 2)
df_last.info()
```

```python
anomaly = df_last['Negative']

anomaly.plot.box(title='anomalys')
plt.show()
```

```python
df_last=drop_outliers_zscore(df_last, 'Positive', 2)
```

```python
anomaly = df_last['Positive']

anomaly.plot.box(title='anomalys')
plt.show()
```

```python
df_need=drop_outliers_zscore(df_need, 'Negative', 2)
```

```python
df_need.loc[:, 'Positive_Ratio'] = df_need['Positive'] / (df_need['Positive'] + df_need['Negative'] + 1)
df_need.head()
```

```python
target_columns = ['AppID', 'Name', 'Release date', 'Developers', 'Genres', 'Positive', 'Negative', 'Positive_Ratio']

df_final = df_need[(df_need['Positive'] + df_need['Negative'] > 0)][target_columns].copy()
df_final.head()
```
